```python
import pandas as pd
import numpy as np

# load the dataset
df = pd.read_csv('WorldCupMatches.csv')
#df.head()
#df.describe()
#print(df)

# to solve the "Symmetric Pair" problem.
# sort the team initials alphabetically row-by-row so ARG-BRA and BRA-ARG are always recorded in the exact same order (Team1, Team2)
# if we are not do soring, python treats these two tewam ARGandBRA as two completely different matchups.
df['Team1'] = np.where(df['Home Team Initials'] < df['Away Team Initials'], df['Home Team Initials'], df['Away Team Initials'])
df['Team2'] = np.where(df['Home Team Initials'] < df['Away Team Initials'], df['Away Team Initials'], df['Home Team Initials'])

#  group by our new consolidated pairs, count matches, and sum home goals
#  telling Python: "Group the rows by Team1 and Team2, then pick the 'Home Team Initials' column and count how many times it appears for that pair."
team_pairs = df.groupby(['Team1', 'Team2']).agg(
    weight=('Home Team Initials', 'count'),
    HomeGoalTotal=('Home Team Goals', 'sum')
).reset_index()

# rename columns
team_pairs = team_pairs.rename(columns={
    'Team1': 'Home Team Initials',
    'Team2': 'Away Team Initials'
})

# display the data to verify
print(team_pairs.head())

#  Write to CSV
team_pairs.to_csv('team_pairs.csv', index=False)
```
